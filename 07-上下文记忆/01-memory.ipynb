# 上下文记忆
## 如何解决记忆问题
### 上下文工程
需要`额外的模块`去保存我们和模型对话的上下文信息，然后下一次请求时，把历史信息都输入给模型，让模型输出结果。

在LangChain中，我们使用`Memory`模块来保存上下文信息。

`上下文工程(Context Engineering)` 负责 `“合理组织”这些记忆和任务信息` ，让LLM的响应更连贯、更贴
合需求。这也是Agent能实现 `复杂多轮交互` 的核心基础。

### 上下文类型及相关的API

LangChain的上下文工程是基于Agent讨论的，而上下文工程是构建在LangGraph之上的。

LangGraph 提供了三种管理上下文的方法，这些方法结合了可变性和生命周期维度:

- 动态运行时上下文： 在单次运行中会演变的可变数据，通过LangGraph的state对象访问。声明周期单次运行，具有可变性。
  
- 动态跨会话上下文：在对话间共享的持久数据。具有动态可变性，访问方式是：LangGraph store对象。声明周期跨对话。
  
- 静态运行上下文，在启动时传入的用户元数据、工具、数据库连接，静态不可变，生命周期是单次运行，通过LangGraph的context对象访问。


## LangChain的记忆
### 记忆的分类
记忆分为短期记忆和长期记忆，对应不同的使用场景：
- 短期记忆（Short-term memory、会话级记忆、thread-scoped memory）：作用范围是单个
对话线程（Thread）内，一旦开启新对话（更换 thread_id ），记忆即消失。
- 长期记忆（Long-term memory，跨会话级记忆 ）：在会话间存储用户特定或应用级数据， 并
在 会话线程间共享 。它可以随时在任何线程中被调用。记忆的范围是任意自定义命名空间，而不
仅仅是单一线程 ID。
### 记忆的管理
LangChain1.x版本，构建在LangGraph图结构之上的，通过上文提到的state和store构
建记忆系统。使用更简单、功能更统一。
state：短期记忆对象，以会话为单位组织，包含当前绘画的所有消息记录以及自定义信息。
store：长期记忆对象，跨会话持久化的数据，通常需要结合向量数据库或外部存储实现。
